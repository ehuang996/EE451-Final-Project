### Encode non-numerical binary/non-binary representations of to using numbers

```python
# One-hot encode the categorical columns
data_train = pd.get_dummies(data_train, columns=['map'], drop_first=False)
data_test = pd.get_dummies(data_test, columns=['map'], drop_first=False)
```

```python
encoder = LabelEncoder()

# Maps
data_train['map_de_cache'] = encoder.fit_transform(data_train['map_de_cache'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_dust2'] = encoder.fit_transform(data_train['map_de_dust2'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_inferno'] = encoder.fit_transform(data_train['map_de_inferno'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_mirage'] = encoder.fit_transform(data_train['map_de_mirage'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_nuke'] = encoder.fit_transform(data_train['map_de_nuke'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_overpass'] = encoder.fit_transform(data_train['map_de_overpass'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_train'] = encoder.fit_transform(data_train['map_de_train'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_train['map_de_vertigo'] = encoder.fit_transform(data_train['map_de_vertigo'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_cache'] = encoder.fit_transform(data_test['map_de_cache'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_dust2'] = encoder.fit_transform(data_test['map_de_dust2'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_inferno'] = encoder.fit_transform(data_test['map_de_inferno'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_mirage'] = encoder.fit_transform(data_test['map_de_mirage'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_nuke'] = encoder.fit_transform(data_test['map_de_nuke'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_overpass'] = encoder.fit_transform(data_test['map_de_overpass'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_train'] = encoder.fit_transform(data_test['map_de_train'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data_test['map_de_vertigo'] = encoder.fit_transform(data_test['map_de_vertigo'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

# Round winner
round_winner_mappings = {'CT': 1, 'T': -1}
data_train['round_winner'] = data_train['round_winner'].map(round_winner_mappings)
data_test['round_winner'] = data_test['round_winner'].map(round_winner_mappings)

# Bomb planted
data_train['bomb_planted'] = encoder.fit_transform(data_train['bomb_planted'])
bomb_planted_mappings = {index : label for index, label in enumerate(encoder.classes_)}
data_test['bomb_planted'] = encoder.fit_transform(data_test['bomb_planted'])
bomb_planted_mappings = {index : label for index, label in enumerate(encoder.classes_)}

print(map_mappings)
print(round_winner_mappings)
print(bomb_planted_mappings)
```
