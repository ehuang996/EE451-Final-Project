#  This file is for data cleaning of the csgo_round_snapshots.csv

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv('csgo_round_snapshots.csv')
data
```

```python
pd.set_option('display.max_rows', None)
np.sum(data.isnull())
```

### Encode non-numerical binary/non-binary representations of to using numbers

```python
encoder = LabelEncoder()
data['map'] = encoder.fit_transform(data['map'])
map_mappings = {index : label for index, label in enumerate(encoder.classes_)}

data['round_winner'] = encoder.fit_transform(data['round_winner'])
round_winner_mappings = {index : label for index, label in enumerate(encoder.classes_)}


data['bomb_planted'] = encoder.fit_transform(data['bomb_planted'])
bomb_planted_mappings = {index : label for index, label in enumerate(encoder.classes_)}

print(map_mappings)
print(round_winner_mappings)
print(bomb_planted_mappings)
```

## Save to folder

```python
# Save data to local folder with a new name
data.to_csv('csgo_round_snapshots_cleaned.csv', index=False)
```
